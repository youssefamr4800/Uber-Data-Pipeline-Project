### **SPARK STREAMING**

```python
entities = ['customers', 'trips', 'locations', 'payments', 'vehicles', 'drivers']
for entity in entities:
    df_batch = spark.read.format('csv')\
                .option('header', True)\
                .option('inferSchema', True)\
                .load(f'/Volumes/uber_dataset_project/source/source_data/{entity}/')

    entity_schema = df_batch.schema

    df = spark.readStream.format('csv')\
                .option('header', True)\
                .schema(entity_schema)\
                .load(f'/Volumes/uber_dataset_project/source/source_data/{entity}/')

    df.writeStream.format('delta')\
                .outputMode('append')\
                .option('checkpointLocation', f'/Volumes/uber_dataset_project/bronze/checkpoints/{entity}')\
                .trigger(once = True)\
                .toTable(f'uber_dataset_project.bronze.{entity}')
```
